```python
import pandas as pd

students = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="students")

marks = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="marks")

contact = pd.read_excel("Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="contact")

#1. INNER MERGE
inner_merge = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="inner"
)

print(inner_merge.head())
print(inner_merge.shape)
#2. LEFT MERGE
left_df = pd.merge(
    students,
    marks,
    on = "Student_ID",
    how = "left"
)
print(left_df.tail())
#Even if marks are missing,students remain.
#3. RIGHT MERGE

right_merge = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="right"
)

print(right_merge.head())
#4. OUTER MERGE

outer_merge = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="outer"
)
print(outer_merge.tail())
```

### Pivot Files

```python
"""A Pivot table is used to summarize and analyse data by grouping rows and applaying aggregate functions such as sum, mean, count, max,min etc."""
#step 1: load dataset
import pandas as pd
df = pd.read_csv("people-100.csv")
print(df.head())
# step 2 : Create age column
df['Date of birth']  = pd.to_datetime(df['Date of birth'])
today = pd.Timestamp.today()
df['Age'] = ((today - df["Date of birth"]).dt.days // 365)
print(df[["First Name","Sex","Age"]].head())
```

```python
# 1.Basic pivot 
# Average age by gender
pivot1 = pd.pivot_table(
    df,
    values = "Age",
    index = "Sex"
)
print(pivot1)
# 2. Pivot table with aggfunc = 'mean'
pivot2 = pd.pivot_table(
    df,
    values = "Age",
    index = "Sex",
    aggfunc = 'mean'
)
print(pivot2)
# 3. Pivot table with aggfunc = 'sum'
pivot3 = pd.pivot_table(
    df,
    values = "Age",
    index = "Sex",
    aggfunc = 'sum'
)
print(pivot3)
# 4. Pivot table with aggfunc = 'count'
pivot4 = pd.pivot_table(
    df,
    values = "Age",
    index = "Sex",
    aggfunc = 'count'
)
print(pivot4)
# 5. Multiple aggregate function
pivot5 = pd.pivot_table(
    df,
    values = "Age",
    index = "Sex",
    aggfunc = ['mean','max','min']
)
print(pivot5)
# 6. Adds overall average
pivot6 = pd.pivot_table(
    df,
    values = "Age",
    index = "Sex",
    aggfunc = 'mean',
    margins = True
)
print(pivot6)
# 7. Group by Job title and gender
# Average age by job title and gender
pivot7 = pd.pivot_table(
    df,
    values = "Age",
    index = "Job Title",
    columns = "Sex",
    aggfunc = 'mean',
)
print(pivot7)
# Count People by John and Gender
pivot8 = pd.pivot_table(
    df,
    values = "User Id",
    index = "Job Title",
    columns = "Sex",
    aggfunc = "count",
    fill_value = 0
)
print(pivot8)
#nan fill value with 0
```

```python
import pandas as pd
df = pd.read_excel("Data_Cleaning_50Plus_Records.xlsx")
# Count duplicates
print(df.duplicated().sum())
```

```python
# 5. Dropping Duplicate records
#before
print(df.shape)
#remove duplicates
df.drop_duplicates(inplace = True)
#After
print(df.shape)
```

```python
# 6. String operations
# String methods work only on text columns
# Remove Extra space
df['Student_Name'] = df['Student_Name'].str.strip()
# Convert to Uppercase
df['Student_Name'] = df['Student_Name'].str.upper()
print(df['Student_Name'].head())
#Convert to title case
df['Student_Name'] = df['Student_Name'].str.title()
print(df['Student_Name'].head())

#Replace text
#replace department name
df['Department'] = df['Department'].replace({
    'CSE':'Computer Science',
    'ECE': 'Electronics and communi eng',
    'EEE': 'electircal and electronic',
    'IT': 'Information tech',
    'MECH' : 'Mechanical'
})
print(df['Department'].head())
print(df.columns)
```
