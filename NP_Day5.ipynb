```python
#Figure and axes
#These are the most important concepts
#Figure- The entire chart window
#axes the area where the grapy is drown
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,30,40,50]
fig,ax = plt.subplots()
ax.plot([1,2,3],[10,20,30])
#Adding a title
plt.title("Monthly sales")
#Adding x and y labels
plt.xlabel("Month")
plt.ylabel("Sales")
#chaning line color
plt.plot(x,y,color = 'blue')
#Chaning line style
plt.plot(x,y,linestyle = '-.')
#changing the line width
plt.plot(x,y,linewidth = 4)
#Thicker line
#Adding markers
plt.plot(x,y,marker = 's')
```

```python
"""Box plot
A box plot is used to visualize the distribution of numerical data and identify outliers.
components of a box plot
Min--- Q1---[Median]---Q3-----Max
    |                      |
Lower whisker        Upper Whisker
Minimum:Smallest value(excluding outliers)
Q1(25%): First quartile
Median (50%): Middle value
Q3(75%): Third quartile
Maximum: Largest value(excluding outliers)
Outliers: Values far from the rest of the data, shown as individual points
"""
```

```python
import matplotlib.pyplot as plt
salary = [25000,2800,30000,50000,45000,350000,67000,70000,100000]
plt.boxplot(salary)
plt.title("Salary Distrubution")
plt.ylabel("Salary")
plt.show()
```

```python
"""Heatmap what is a heatmap?
 A heatmap uses colors to represent values in a table or matrix.
 In data analysis, heatmaps are commonly used to visualize correlations between numerical columns.

 correlation values
 Value meaning
 +1  -> Strong Positive Relationship
 0   -> No Relationship
 -1  -> Strong Negative Relationship
 """
#Correlation heatmap with seaborn
import seaborn as sns
import matplotlib.pyplot as plt
#load dataset
tips = sns.load_dataset("tips")
#correlation matrix
corr = tips.corr(numeric_only = True)
#Heatmap
sns.heatmap(corr,annot = True)
plt.title("Correlation heatmap")
plt.show()
```

```python
#Histogram using seaborn
sns.histplot(
    data=tips,
    x="total_bill",
    bins=40
)
plt.title("distribution of total bill")
plt.show()
#sns,histplot()
#creates histogram
```

```python
'''KDE Plot (Kernal Density Estimation) what is KDE plot?
A KDE plot is a smooth version of a histogram
Instead of bars,it shows a smooth curve representing data distribution
When to use KDE? Understanding data distribution identifying peaks comparing distribution statistical analysis'''
sns.kdeplot(
    data=tips,
    x="total_bill"
)
plt.title("KDE Plot")
plt.show()
```

```python
#histogram +KDE together
#most commonly used in data science
sns.histplot(
    data=tips,
    x="total_bill",
    kde=True
)
plt.title("histogram with KDE")
plt.show()
```
