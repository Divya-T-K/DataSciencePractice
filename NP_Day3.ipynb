### Series in pandas

```python
import pandas as pd
a = pd.Series([1,2,3,4,6])
b = pd.Series([1.5,True,3.8,4.2,6])
print(a)
print(a.values)
print(a.index)
print(a.dtype)
print(b.dtype)
#accessing elements
print(a[1])
#access elements using loc
print(a.iloc[1])
#slicing
print(a[2:4])
print(a.loc[1:3])
print(a.iloc[2:5])
```

```python
import pandas as pd
arr = pd.Series([1,2,3,4,56,2],index = ['A','B','C','D','E','F'])
print(arr[['A','C','F']])
print(arr)
print(arr['E'])
print(arr + 5)
print(arr.iloc[2]) #position based accessing
print(arr.loc['A']) #labeled value based accessing
print(arr.sum())
print(arr.std())
print(arr.min())
print(arr.max())
print(arr.mean())
print(arr.var())
print(arr.median())
# boolean indexing
print(arr[arr>30])
print(len(arr))
#it return individual value count
print(arr.value_counts())
#string operations
name = pd.Series(['tharunika','diya','uthra','ram','gokul'])
print(name.str.upper())
print(name.str.len())
```

### Data Frames in pandas

```python
import pandas as pd
# method1 : from dictionary
data = {'name':['Diya','Tharunika','Uthra'], 'rollno':[11,53,23], 'Age':[18,20,19]}
df = pd.DataFrame(data)
print(df)
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([5.6,9.54,9.4,54.6,6])
s3 = pd.Series([True,False,True,False])
df = pd.DataFrame({'sno':s1,'cgpa':s2,'attendance':s3})
print(df)

#list of list
data = [['Diya',18,98],['Tharun',19,96],['Rashvandh',19,82]]
df = pd.DataFrame(data)
print(df)
df = pd.DataFrame(data,columns=["Name","Age","Mark"])
print(df)

#list of dict
data = [{"Name":"Diya","Course":"AIML","Grade":"B"},{"Name":"Tharun","Course":"AIDS","Grade":"A"}]
df = pd.DataFrame(data,index=["std1","std2"])
print(df)

#adding one extra columns
df['marks'] = [98,96]
print(df)

#adding one extra row
std3 = pd.DataFrame([{"Name":"Rash","Course":"AIML","Grade":"B"}])
df = pd.concat([df,std3])
print(df)

#creating multiple rows
std = pd.DataFrame([{"Name":"Uthra","Course":"AIML","Grade":"A++","Mark":99},
                    {"Name":"Dharshini","Course":"AIDS","Grade":"B","Mark":80}])
df = pd.concat([df,std])
print(df)

#Column access by bracket notation and dot notation
print(df.Course)
print(df[["Grade","Mark"]])
print(df.shape)
print(df.shape[0])
print(df.shape[1])
print(df.dtypes)
print(df.info())
print(df.describe())
# select specific row
print(df.loc[['std1','std2']])
print(df.iloc[3])
#updatevalue
df.loc[0,'Grade'] = "B++"
print(df)
# Delete column
df.drop("Grade",axis = 1,inplace = True)
print(df)
# Delete row
df.drop("std2",axis = 0,inplace = False)
print(df)
#sorting
df.sort_values('Mark',inplace = True)
print(df)
df.sort_values('Mark', ascending = False,inplace = True)
print(df)
#statictical functio for particular column
print(df['Mark'].max())
```
