```python
import pandas as pd
from scipy.stats import zscore

# Load dataset
df = pd.read_csv("Messy_Employee_dataset.csv")

# Select numerical columns
numeric_df = df.select_dtypes(include='number')

# -----------------------------------
# 1. Descriptive Statistics
# -----------------------------------
descriptive_stats = numeric_df.describe()

print("Descriptive Statistics:")
print(descriptive_stats)

# -----------------------------------
# 2. Identify Skewed Distributions
# -----------------------------------
skewness = numeric_df.skew()

print("\nSkewness Values:")
print(skewness)

for col, value in skewness.items():
    if value > 0.5:
        print(f"{col}: Right-Skewed")
    elif value < -0.5:
        print(f"{col}: Left-Skewed")
    else:
        print(f"{col}: Approximately Symmetric")

# -----------------------------------
# 3. Z-Score Normalization
# -----------------------------------
zscore_df = numeric_df.apply(
    lambda x: zscore(x, nan_policy='omit')
)

print("\nZ-Score Normalized Data:")
print(zscore_df.head())

# -----------------------------------
# 4. Pearson Correlation
# -----------------------------------
pearson_corr = numeric_df.corr(method='pearson')

print("\nPearson Correlation Matrix:")
print(pearson_corr)

# -----------------------------------
# 5. Spearman Correlation
# -----------------------------------
spearman_corr = numeric_df.corr(method='spearman')

print("\nSpearman Correlation Matrix:")
print(spearman_corr)

# -----------------------------------
# 6. Save Results
# -----------------------------------
with pd.ExcelWriter("Statistical_Analysis.xlsx") as writer:
    descriptive_stats.to_excel(writer, sheet_name="Descriptive_Stats")
    skewness.to_frame("Skewness").to_excel(writer, sheet_name="Skewness")
    zscore_df.to_excel(writer, sheet_name="ZScore_Normalized", index=False)
    pearson_corr.to_excel(writer, sheet_name="Pearson_Correlation")
    spearman_corr.to_excel(writer, sheet_name="Spearman_Correlation")

print("\nAnalysis saved successfully!")
```

```python
# Check Missing values
import pandas as pd
import seaborn as sns
from sklearn.impute import KNNImputer
# Load Titanic dataset
df = sns.load_dataset("titanic")
# forward fill for embarked
df['embarked_ffill'] = df['embarked'].ffill()
#KNN Imputation for age
cols = ['age','fare','sibsp','parch','pclass']
imputer = KNNImputer(n_neighbors = 5)
df[cols] = imputer.fit_transform(df[cols])
print(df.isnull().sum())
```

```python
# Backward fill
# df.fillna(method = 'bfill',inplace = True)
"""Encoding:
-> Machine Learning models work with numbers,not text
ex: sex
  -> male,female,male
  -> A models cannot directly understand 'male' and 'female'
  -> We convert categories into numbers encoding
types:
1. Label Encoding(ordinal data)
2. One hot encoding (norminal data)
3. Targeted encoding
"""
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
# Load Titanic dataset
df = sns.load_dataset('titanic')
# Label Encoding
le = LabelEncoder()
df['sex_encoded'] = le.fit_transform(
    df['sex']
)
# One-Hot Encoding
df = pd.get_dummies(df,columns = ['embarked'],drop_first = True)
# Target Encoding
target_map = df.groupby('pclass')['survived'].mean()
df['class_target'] = (df['pclass'].map(target_map))
print(df.head())
```

```python
#Feature Scaling
''' >> transforms numerical features to a similar scale
>> Many ML algorithms calculate distance,so fare can dominate the model
# METHOD1: MiniMaxScaler --->sensitive to outliers , values between 0 and 1
# METHOD2: StandardScaler --->here outlier is applicable , works well for normally distributed data
Best for : Logistic Regression,SVM,PCA,KNN,Linear Regression
# METHOD3: RobustScaler  ---->uses Median and IQR instead of Mean and SD,handels outliers well,values not limited to 0-1
'''
# 1. MinMaxScaler
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[['age_minmax','fare_minmax']] = scaler.fit_transform(
    df[['age','fare']]
)
df[['age_minmax','fare_minmax']].head()
```

```python
# MinMax Chart
import matplotlib.pyplot as plt
plt.hist(df['fare'].dropna(),bins = 30)
plt.title("Original Fare")
plt.show()
plt.hist(df['fare_minmax'], bins = 30)
plt.title("MinMax Scaled Fare")
plt.show()
```

```python
# 2. Standard Scaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df[['age_std','fare_std']] = scaler.fit_transform(
    df[['age','fare']]
)
df[['age_std','fare_std']].head()
```

```python
# 3. Robust Scaler
from sklearn.preprocessing import RobustScaler
scaler = RobustScaler()
df[['age_robust','fare_robust']] = scaler.fit_transform(
    df[['age','fare']]
)
df[['age_robust','fare_robust']].head()
```
