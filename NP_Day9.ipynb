## MACHINE LEARNING

### Machine Learning (ML)
#### 1. What is Machine Learning?

Machine Learning is a branch of Artificial Intelligence (AI) that enables computers to learn from data and make predictions or decisions without being explicitly programmed.

##### Simple Definition

Machine Learning is the process of teaching a computer to learn patterns from data and improve its performance automatically.

##### Example
- Netflix recommends movies based on your watching history.
- Gmail detects spam emails.
- Amazon suggests products you may like.


#### 2. Process of Machine Learning
##### Step-by-Step Process
1. Collect Data
   > Gather data from different sources.
2. Prepare Data
   > Clean missing values and remove errors.
3. Select Features
   > Choose important columns for learning.
4. Split Data
   > Divide data into Training and Testing sets.
5. Choose Algorithm
   > Select a suitable ML algorithm.
6. Train the Model
   > The model learns patterns from training data.
7. Test the Model
   > Check performance using test data.
8. Evaluate Performance
   > Measure accuracy and errors.
9. Deploy the Model
   > Use the model in real-world applications.
10. Monitor and Improve
   > Update the model with new data.
#### 3. Types of Machine Learning

There are mainly **4 types of Machine Learning**:

##### 1. Supervised Learning
- Data contains input and correct output labels.
- Model learns from labeled data.

##### Example:

- Predicting house prices
- Email spam detection
##### 2. Unsupervised Learning
- Data has no labels.
- Model finds hidden patterns.

##### Example:
- Customer segmentation
- Grouping similar products
##### 3. Semi-Supervised Learning
- Uses both labeled and unlabeled data.

##### Example:

- Image classification with few labeled images
##### 4. Reinforcement Learning
- Model learns through rewards and penalties.

##### Example:

- Self-driving cars
- Game-playing AI

#### 7. Underfitting and Overfitting
#### Underfitting
#### Definition

Underfitting happens when the model is too simple and cannot learn the patterns in the data properly.

#### Characteristics
- Low training accuracy
- Low testing accuracy
- High error
#### Example

Trying to fit a straight line to data that actually follows a curve.

#### Solution
- Use a more complex model
- Add more features
- Train longer
#### Overfitting
#### Definition

Overfitting happens when the model learns the training data too well, including noise and unnecessary details.

#### Characteristics
- Very high training accuracy
- Low testing accuracy
- Poor performance on new data
#### Example

A student memorizes answers instead of understanding concepts.

#### Solution
- Use more training data
- Reduce model complexity
- Apply regularization
- Use cross-validation
